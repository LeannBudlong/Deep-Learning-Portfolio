## **Concept Review**

### Forward Propagation

Forward propagation is the process of passing input values through a neural network to produce an output or prediction.

The flow of information is:

<p align="center"><b>Input Layer → Hidden Layer → Output Layer</b></p>

Each neuron computes a weighted sum using:

$$
Z = \sum (w_i x_i) + \theta
$$

where:

- $x_i$ = input value  
- $w_i$ = corresponding weight  
- $\theta$ = bias  
- $Z$ = weighted sum before activation  



### ReLU Activation Function

This laboratory uses the **Rectified Linear Unit (ReLU)** activation function:

$$
f(Z) = \max(0, Z)
$$

This means:

- if $Z > 0$, the output is $Z$
- if $Z \leq 0$, the output becomes $0$

ReLU allows positive values to pass through while removing negative values.


### Output and Error

After the hidden layer outputs are passed to the output neuron, the network produces a predicted value:

$$
\hat{y}
$$

The prediction is then compared with the actual target value:

$$
y
$$

The error is calculated as:

$$
\delta = \hat{y} - y
$$

This error shows the difference between the network's prediction and the expected output.


## **Implementation**

#### **Import Libraries**

```python
import numpy as np
```

#### **Define the Inputs and Network Parameters**

The network contains three input values, two hidden neurons, and one output neuron.

```python
# Input and actual output
x = np.array([1, 0, 1])
y = np.array([1])

# Hidden layer weights
W_hidden = np.array([
    [0.2, -0.3],
    [0.4,  0.1],
    [-0.5, 0.2]
])

# Output layer weights
W_output = np.array([
    [-0.3],
    [-0.2]
])

# Bias values
theta_hidden = np.array([-0.4, 0.2])
theta_output = 0.1

print("Input:", x)
print("Actual Output:", y)
```

#### **Define the ReLU Activation Function**

As reviewed earlier, the ReLU function prevents negative weighted sums from being passed to the next layer.

```python
def relu(z):
    return np.maximum(0, z)
```

#### **Compute the Hidden Layer**

***Hidden Layer 1***

$$
Z_1 = (w_{11}x_1) + (w_{13}x_2) + (w_{15}x_3) + \theta_1
$$

Substituting the given values:

$$
Z_1 = (0.2)(1) + (0.4)(0) + (-0.5)(1) + (-0.4)
$$
$$
Z_1 = -0.7
$$

Applying the ReLU activation function:

$$
H_1 = \max(0, -0.7)
$$

$$
\boxed{H_1 = 0}
$$

***Hidden Layer 2***
$$
Z_2 = (w_{12}x_1) + (w_{14}x_2) + (w_{16}x_3) + \theta_2
$$

Substituting the given values:

$$
Z_2 = (-0.3)(1) + (0.1)(0) + (0.2)(1) + (0.2)
$$

$$
Z_2 = 0.1
$$

Applying the ReLU activation function:

$$
H_2 = \max(0, 0.1)
$$

$$
\boxed{H_2 = 0.1}
$$

```python
z_hidden = np.dot(x, W_hidden) + theta_hidden
h = relu(z_hidden)

print("Hidden Weighted Sums:", z_hidden)
print("Hidden Layer Outputs:", h)
```

#### **Compute the Output Layer**

The weighted sum of the output neuron is:

$$
Z_3 = (w_{21}H_1) + (w_{22}H_2) + \theta_3
$$

Substituting the hidden layer outputs:

$$
Z_3 = (-0.3)(0) + (-0.2)(0.1) + (0.1)
$$


$$
Z_3 = 0.08
$$

Applying the ReLU activation function:

$$
\hat{y} = \max(0, 0.08)
$$

$$
\boxed{\hat{y} = 0.08}
$$


```python
z_output = np.dot(h, W_output).item() + theta_output
y_hat = relu(z_output)

print("Output Weighted Sum:", z_output)
print("Predicted Output:", y_hat)
```

#### **Compute the Error**

$$
y = 1, \qquad \hat{y} = 0.08
$$

Using:

$$
\delta = \hat{y} - y
$$

Therefore:

$$
\delta = 0.08 - 1 = \boxed{-0.92}
$$

```python
error = y_hat - y[0]

print("Actual Output:", y[0])
print("Predicted Output:", y_hat)
print("Error:", error)
```

## **Results and Discussion**

The forward pass produced hidden-layer outputs of:

$$
H_1 = 0, \qquad H_2 = 0.1
$$

Using these values, the output neuron produced:

$$
\hat{y} = 0.08
$$

Compared with the actual output:

$$
y = 1
$$

the error is:

$$
\delta = \hat{y} - y = -0.92
$$

The forward pass shows how the network transforms the original inputs into a prediction through weighted connections and activation functions. The first hidden neuron produced a negative weighted sum, so ReLU reduced its output to zero, meaning that neuron did not contribute to the final prediction. Only the second hidden neuron remained active, producing a relatively small contribution to the output layer.

## **Conclusion**



As a result, the network predicted only $0.08$ when the target value was $1$. The error of $-0.92$ therefore indicates that the current set of weights and biases causes the network to strongly underestimate the expected output. This means that the network has not yet learned parameter values that can produce an output close to the target. In a training process, this error would be used during backpropagation to determine how the weights and biases should be adjusted.
